## Customers per country

```python
country_counts = customers["country"].value_counts()
country_counts
```

```python
plt.figure(figsize=(10,6))
country_counts.plot(kind = "bar")
plt.xlabel("Country")
plt.ylabel("Number of Customers")
plt.title("Customers by Country")

plt.show()
```

## Customers per Ages

```python
bins = [0, 9, 19, 29, 39, 49, 59, 69, 79, 100]
labels = ["0-9", "10-19", "20-29", "30-39", "40-49",
          "50-59", "60-69", "70-79", "80-100"]

customers["age_group"] = pd.cut(customers["age"], bins=bins, labels=labels)
age_counts = customers["age_group"].value_counts().sort_index()
age_counts
```

```python
plt.figure(figsize=(10,6))
age_counts.plot(kind = "bar")
plt.xlabel("Age Groups")
plt.ylabel("Number of Customers")
plt.title("Customers by Age")

plt.show()
```

## Spends by age group

```python

age_spend = customers.groupby("age_group")["total_spend_usd"].agg(
    total_spend="sum",
    avg_spend="mean",
    customer_count="count"
).reset_index()

age_spend
```

```python
plt.figure(figsize=(8,5))
plt.bar(age_spend["age_group"], age_spend["total_spend"])
plt.title("Total Spend by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Total Spend (USD)")
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(age_spend["age_group"], age_spend["avg_spend"])
plt.title("Average Spend per Customer by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Spend (USD)")
plt.show()
```

## Male vs Female spending

```python
gender_summary = customers.groupby("gender").agg(
    counts=("customer_id", "count"),
    total_revenue=("total_spend_usd", "sum"),
    avg_spend_per_customer=("total_spend_usd", "mean"),
    avg_orders=("total_orders", "mean"),
    avg_order_value=("avg_order_value_usd", "mean")
).reset_index()

gender_summary
```

```python
plt.figure(figsize=(6,4))
plt.bar(gender_summary["gender"], gender_summary["avg_spend_per_customer"])
plt.title("Average Spend per Customer by Gender")
plt.xlabel("Gender")
plt.ylabel("Avg Spend (USD)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(gender_summary["gender"], gender_summary["avg_orders"])
plt.title("Average Number of Orders by Gender")
plt.xlabel("Gender")
plt.ylabel("Avg Total Orders")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(gender_summary["gender"], gender_summary["avg_order_value"])
plt.title("Average Order Value by Gender")
plt.xlabel("Gender")
plt.ylabel("Avg Order Value (USD)")
plt.show()
```

## Spending by Membership Tier

```python
member_summary = customers.groupby("membership_tier").agg(
    counts=("customer_id","count"),
    total_revenue=("total_spend_usd", "sum"),
    avg_spend_per_customer=("total_spend_usd", "mean"),
    total_orders=("total_orders","sum"),
    avg_orders=("total_orders", "mean"),
    avg_order_value=("avg_order_value_usd", "mean")
).reset_index()

tier_order = ["Free", "Silver", "Gold", "Platinum"]

member_summary["membership_tier"] = pd.Categorical(
    member_summary["membership_tier"],
    categories=tier_order,
    ordered=True
)
member_summary = member_summary.sort_values("membership_tier").reset_index(drop=True)

member_summary
```

```python
plt.figure(figsize=(6,4))
plt.bar(member_summary["membership_tier"], member_summary["avg_spend_per_customer"])
plt.title("Average Spend per Customer by Membership Tier")
plt.xlabel("Membership Tier")
plt.ylabel("Avg Spend (USD)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(member_summary["membership_tier"], member_summary["total_revenue"])
plt.title("total_revenue by Membership Tier")
plt.xlabel("Membership Tier")
plt.ylabel("total_revenue")
plt.show()
```

## Spending by channel

```python
channel_summary = customers.groupby("acquisition_channel").agg(
    counts=("customer_id","count"),
    total_revenue=("total_spend_usd", "sum"),
    avg_spend_per_customer=("total_spend_usd", "mean"),
).reset_index()

channel_summary
```

```python
plt.figure(figsize=(6,4))
plt.bar(channel_summary['acquisition_channel'],channel_summary['total_revenue'])
plt.title("Average Spend per Customer by Acquisition Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("Avg Spend (USD)")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.bar(channel_summary['acquisition_channel'],channel_summary['avg_spend_per_customer'])
plt.title("Average Spend per Customer by Acquisition Channel")
plt.xlabel("Acquisition Channel")
plt.ylabel("Avg Spend (USD)")
plt.xticks(rotation=45)
plt.show()
```
